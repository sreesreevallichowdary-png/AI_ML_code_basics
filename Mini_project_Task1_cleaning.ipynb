```python
import json
import pandas as pd
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Step 1: Load JSON file
with open("trends_20260411.json", "r", encoding="utf-8") as f:
    stories = json.load(f)

    # Step 2: Convert to DataFrame
df = pd.DataFrame(stories)
print(df.info)


```

```python
# Inspect & Handle Missing Values_All columns data types
#Display the data types of all columns.
print("Column Names:")
print(df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
```

```python
#Cleaing#Inspect & Handle Missing Values
#Count the number of missing values in each column.
print("Missing Values:")
print(df.isnull().sum())
print(f"\nTotal missing values: {df.isnull().sum().sum()}")
```

```python
#Cleaing
#Duplicate Rows
# Check for duplicates
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

if duplicates > 0:
    print("\nDuplicate rows:")
    print(df[df.duplicated(keep=False)])
```

```python
# Check if categorical columns are properly typed
print("Unique values in categorical columns:\n")

categorical_cols = ['category', 'author','num_comments','collected_at']
for col in categorical_cols:
    print(f"{col}: {df[col].unique()}")
    print(f"  Count: {df[col].nunique()} categories\n")
```
